# Challenge 2: Structuring Data

In this challenge, we will continue to work with customer data from an insurance company, but we will use a dataset with more columns, called marketing_customer_analysis.csv, which can be found at the following link:

https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv

This dataset contains information such as customer demographics, policy details, vehicle information, and the customer's response to the last marketing campaign. Our goal is to explore and analyze this data by performing data cleaning, formatting, and structuring.

```python
# 
df_market = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv")
df_market.head()
```

```python
df_market['effective_to_date']=pd.to_datetime(df_market['effective_to_date'])
# Clean column names
df_market.columns = df_market.columns.str.lower().str.replace(' ', '_')

# Convert revenue to numeric (if not already)
df_market['total_claim_amount'] = pd.to_numeric(df_market['total_claim_amount'], errors='coerce')
```

```python
from cleaning_functions import quick_data_report
quick_data_report(df_market)
```

1. You work at the marketing department and you want to know which sales channel brought the most sales in terms of total revenue. Using pivot, create a summary table showing the total revenue for each sales channel (branch, call center, web, and mail).
Round the total revenue to 2 decimal points.  Analyze the resulting table to draw insights.

```python
df_market['sales_channel'].head()
df_market['total_claim_amount'].head()
```

```python
# Pivot the DataFrame to see the total sales based on sales channel and year
pivot_df = df_market.pivot_table(
    index='sales_channel',
    values=['total_claim_amount'],
    aggfunc='sum'
).round(2)

#Sort values decending
pivot_df.sort_values(by='total_claim_amount', ascending=False)

#Insights: Agent sales channel is the best seller among the other way  of sales.  Branch is following it. In this point if you want to sell more you need agents.
```

2. Create a pivot table that shows the average customer lifetime value per gender and education level. Analyze the resulting table to draw insights.

```python
#  Create pivot table see the customer_lifetime_value based on gender and education level.
pivot_clv_df = df_market.pivot_table(
    index = ['gender'],
    columns = ['education'],
    values = ['customer_lifetime_value'],
    aggfunc= 'mean')
    
pivot_clv_df
```

### Analysing
#### Key Insights:
##### Education Level Effects:
For both genders, "High School or Below" tends to have the highest or one of the highest CLV values, which is surprising — one might expect higher education to correlate with higher spending.

Doctors consistently have the lowest CLV across genders, which may suggest less engagement or fewer purchases despite potentially higher income.
##### Gender Differences:
Female customers with "High School or Below" education have the highest CLV overall at \$8,675.22.

Male customers are more consistent across education levels, with Master's and College educated males having the highest CLV values.

Males with College education show stronger CLV than females with the same education.

### Business Implications:
Marketing efforts targeting less-educated female customers may yield higher lifetime value — possibly due to loyalty or specific product interest.

There's potential to re-engage Doctor-level customers, possibly with targeted campaigns, as they currently underperform.

Consider segmenting campaigns not just by income or education but by combined demographic behavior.


## Bonus

You work at the customer service department and you want to know which months had the highest number of complaints by policy type category. Create a summary table showing the number of complaints by policy type and month.
Show it in a long format table.

*In data analysis, a long format table is a way of structuring data in which each observation or measurement is stored in a separate row of the table. The key characteristic of a long format table is that each column represents a single variable, and each row represents a single observation of that variable.*

*More information about long and wide format tables here: https://www.statology.org/long-vs-wide-data/*

```python
# highest complaints by policy type each month
# the number of complaints by policy type and month
#  Create pivot table 
pivot_complaints = df_market.pivot_table(
    index = ['months_since_policy_inception'],
    columns = ['policy_type'],
    values = ['number_of_open_complaints'],
    )
    
pivot_complaints
```

```python
long_format = pivot_complaints.stack(future_stack=True).reset_index()
long_format.columns = ['months_since_policy_inception', 'policy_type', 'number_of_open_complaints']
long_format
```
